```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
## Data
tapflow_max = 0.2 #kg/s (da simul) 
Q_cond_max = 130 #kW 
m_H2_max = 0.1 #kg/s (dato)

# Enthalpic content of HX
deltaH = Q_cond_max / tapflow_max #kJ/kg
print(f"Enthalpic content of HX: {deltaH:.2f} kJ/kg")

# Proportionality constant 
k = m_H2_max / Q_cond_max #kg_H2/kJ
print(f"Proportionality constant k: {k:.4f} kg_H2/kJ")

# tapflow vs. m_H2 
tapflow = lambda m_H2: 1/(k*deltaH) * m_H2

# plot 
m_H2 = np.linspace(0, m_H2_max, 100)
plt.figure(figsize=(8, 6))
plt.plot(m_H2, tapflow(m_H2), label='Tapflow vs. m_H2', color='blue')
plt.title('Tapflow vs. Mass Flow Rate of H2')
plt.xlabel('Mass Flow Rate of H2 (kg/s)')
plt.ylabel('Tapflow (kg/s)')
plt.axhline(tapflow_max, color='red', linestyle='--', label='Max Tapflow')
plt.legend()
plt.grid()
plt.xlim(0, m_H2_max)
plt.ylim(0, tapflow_max * 1.1)
plt.tight_layout()
plt.show()
```
