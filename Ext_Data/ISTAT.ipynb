```python
import pandas as pd

# Load the list of municipalities of interest..
muni_list = pd.read_csv(
    "C:/Users/Simo/Documents/GitHub/INRE/comuni_varese.csv",
    header=None,
    names=["Comune"],
    dtype=str,
)

# Load the ISTAT data.
istat = pd.read_csv(
    "C:/Users/Simo/Documents/GitHub/INRE/istat_data.csv",
    quotechar='"',
    sep=';',
    skiprows=1,
    skipfooter=1,
    engine='python',
    dtype={'Comune': str, 'Totale': float}
)

# only keep the rows where comune is one of the municipalities in muni_list
istat_2 = istat[istat["Comune"].str.upper().isin(muni_list["Comune"].str.upper())]
# Sum the Totale column for the municipalities of interest
istat_sum = istat_2[istat_2["Età"] == 999]["Totale"].sum()
print(f"Totale per i comuni di interesse: {istat_sum:.2f}")
```
